## **Section 1: Your First Python Program**

The traditional introductory program demonstrates basic output functionality using Python's built-in print function.

````python
# Traditional introductory program demonstrating print function
print("Hello World")
````

## **Section 2: Variables and Basic Data Types**

Python's dynamic typing system automatically infers variable types, eliminating explicit declarations. Variables serve as named references to objects in memory.

````python
# String variable - sequence of Unicode characters
name = "Alice"

# Integer variable - whole number with arbitrary precision
age = 25

# Float variable - IEEE 754 double-precision floating point
height = 1.65

# Boolean variable - logical truth value (True or False)
is_student = True

# List variable - ordered, mutable collection of elements
hobbies = ["reading", "cycling", "photography"]
````

## **Section 3: String Formatting and Output**

Modern Python employs f-string formatting for readable and efficient string interpolation, introduced in Python 3.6.


````python
# String formatting using f-strings with embedded expressions
print(f"Hello, my name is {name}. I am {age} years old and {height} metres tall.")

# Conditional expression within f-string formatting
print(f"I am {'a student' if is_student else 'not a student'}.")

# Method chaining within f-string for list joining
print(f"My hobbies are: {', '.join(hobbies)}.")
````

## **Section 4: Arithmetic Operations**

Python provides comprehensive arithmetic operators following standard mathematical precedence rules and operator associativity.


````python
# Multiple assignment using tuple unpacking
x, y = 10, 3

# Addition operation
print(f"{x} + {y} = {x + y}")

# Subtraction operation
print(f"{x} - {y} = {x - y}")

# Multiplication operation
print(f"{x} * {y} = {x * y}")

# True division - always returns float result
print(f"{x} / {y} = {x / y}")

# Floor division - returns integer quotient
print(f"{x} // {y} = {x // y}")

# Modulus operation - returns remainder after division
print(f"{x} % {y} = {x % y}")

# Exponentiation operation
print(f"{x} ** {y} = {x ** y}")
````

## **Section 5: Built-in Mathematical Functions**

Python's built-in functions provide optimised implementations for common mathematical operations without requiring module imports.

````python
# Floating point value for demonstration
pi = 3.14159
negative_number = -10
numbers = [10, 5, 25, 1]

# Built-in round function with precision specification
print(f"Rounded pi to 2 decimal places: {round(pi, 2)}")

# Built-in abs function to get the absolute value
print(f"Absolute value of {negative_number}: {abs(negative_number)}")

# Built-in max function to find the maximum value in a sequence
print(f"Maximum value in {numbers}: {max(numbers)}")

# Built-in min function to find the minimum value in a sequence
print(f"Minimum value in {numbers}: {min(numbers)}")

# Using constants from the math module
import math
print(f"Value of pi from math module: {math.pi}")
print(f"Value of e from math module: {math.e}")
````

## **Section 6: Conditional Statements - Simple If-Else**

Conditional statements enable branching program execution based on boolean expression evaluation.

````python
# Simple conditional execution
age = 18

# Basic if-else structure for binary decision
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
````

## **Section 7: Multi-way Conditional Statements**

Extended conditional structures handle multiple mutually exclusive conditions using elif clauses.

````python
# Multi-conditional branching demonstration
score = 85

# Cascading conditional evaluation
if score >= 90:
    print("Excellent! You got an A.")
elif score >= 80:
    print("Good job! You got a B.")
elif score >= 70:
    print("You got a C.")
else:
    print("You need to improve.")
````

## **Section 8: While Loops**

While loops provide condition-controlled iteration, executing code blocks repeatedly whilst a boolean condition remains true.

````python
# Counter initialisation for loop control
count = 1

# Condition-controlled iteration
while count <= 5:
    print("Count:", count)
    # Increment using augmented assignment
    count += 1
````

## **Section 9: For Loops with Sequences**

For loops enable iteration over sequences, executing code blocks for each element in the iterable object.

````python
# List creation for iteration demonstration
fruits = ["apple", "banana", "orange"]

# Sequence iteration using for loop
for fruit in fruits:
    print("I like", fruit)
````

## **Section 10: Function Definition and Basic Usage**

Functions encapsulate reusable code blocks, promoting modularity and code organisation through parameterisation and return values.

````python
# Function definition with parameter and return statement
def square(number):
    return number ** 2

# Function invocation with argument passing
result = square(5)

# Output formatting for function result
print("The square of 5 is:", result)
````

````python
# Function definition with multiple inputs and outputs
def calculate_area_perimeter(length, width):
    """Calculates the area and perimeter of a rectangle."""
    area = length * width
    perimeter = 2 * (length + width)
    return area, perimeter

# Function invocation and unpacking results
l = 10
w = 5
area_result, perimeter_result = calculate_area_perimeter(l, w)

# Print the results
print(f"The area of the rectangle with length {l} and width {w} is: {area_result}")
print(f"The perimeter of the rectangle with length {l} and width {w} is: {perimeter_result}")
````

## **Section 11: List Operations and Methods**

Lists provide dynamic, ordered collections with built-in methods for element manipulation and sequence modification.

````python
# List initialisation with string elements
fruits = ["apple", "banana", "orange"]

# Zero-based indexing for element access
print("First fruit:", fruits[0])

# List append method for element addition
fruits.append("grape")
print("Updated fruits list:", fruits)

# List remove method for element deletion
fruits.remove("banana")
print("Final fruits list:", fruits)
````

## **Section 12: Dictionary Operations**

Dictionaries implement key-value mapping using hash tables, providing efficient lookup and modification operations.

````python
# Dictionary creation with initial key-value pairs
person = {
    "name": "John Doe",
    "age": 30,
    "occupation": "Software Engineer"
}

# Dictionary value access using key indexing
print("Name:", person["name"])

# Dictionary modification - adding new key-value pair
person["location"] = "New York"
print("Updated person dictionary:", person)

# Dictionary key deletion using del statement
del person["occupation"]
print("Final person dictionary:", person)
````

## **Section 13: Built-in Functions for Sequences**

Python's built-in function library provides optimised implementations for common sequence analysis and manipulation operations.

````python
# Numeric list for built-in function demonstration
numbers = [5, 2, 8, 1, 9]

# Length calculation using len() function
print("Number of elements:", len(numbers))

# Summation using sum() function
print("Sum of elements:", sum(numbers))

# Maximum value identification using max() function
print("Maximum value:", max(numbers))

# Minimum value identification using min() function
print("Minimum value:", min(numbers))

# Non-destructive sorting using sorted() function
sorted_numbers = sorted(numbers)
print("Sorted list:", sorted_numbers)
````

## **Section 14: Loop Control Statements - Break**

The break statement provides immediate loop termination, transferring control to the statement following the loop construct.

````python
# Loop termination demonstration using break
print("Numbers until 5 is reached:")
for number in range(10):
    # Conditional break statement
    if number == 5:
        break
    print(number)
````

## **Section 15: Loop Control Statements - Continue**

The continue statement skips remaining loop body execution, proceeding directly to the next iteration.

````python
# Loop continuation demonstration using continue
print("Odd numbers only:")
for number in range(10):
    # Skip even numbers using modulus operation
    if number % 2 == 0:
        continue
    print(number)
````

## **Section 16: Nested Loop Structures**

Nested loops enable multi-dimensional iteration patterns, useful for matrix operations and combinatorial problems.

````python
# Nested loop demonstration with conditional output
for i in range(3):
    for j in range(2):
        # Conditional execution within nested structure
        if i == j:
            print("Matching pair:", i, j)
````

## **Section 17: File Opening and Reading**

File I/O operations enable persistent data storage and retrieval through Python's built-in file handling mechanisms.

````python
# File path specification for reading operation
file_path = "data.txt"

# File opening in read mode with explicit closing
try:
    file = open(file_path, "r")
    # Read entire file content into string
    content = file.read()
    print(content)
    # Explicit file closure for resource management
    file.close()
except FileNotFoundError:
    print("File not found:", file_path)
````

## **Section 18: File Writing Operations**

File writing enables program output persistence through various writing modes and content formatting options.

````python
# File creation and writing demonstration
try:
    # Open file in write mode (overwrites existing content)
    file = open("output.txt", "w")
    # Write string content to file
    file.write("Hello, this is a test.")
    # Explicit file closure
    file.close()
    print("File 'output.txt' created successfully")
except Exception as e:
    print("Error writing file:", e)
````

## **Section 19: Context Managers for File Operations**

Context managers provide automatic resource management through the with statement, ensuring proper file closure regardless of exceptions.

````python
# Context manager demonstration for file reading
try:
    # Automatic resource management using with statement
    with open("data.txt", "r") as file:
        content = file.read()
        print("File content using context manager:")
        print(content)
except FileNotFoundError:
    print("Creating example file for demonstration")
    # Context manager for file writing
    with open("example.txt", "w") as file:
        file.write("Example content created by context manager\n")
        file.write("Second line of content\n")
````

## **Section 20: Module Importing**

Python's import system enables code reuse through module inclusion, providing access to external functionality and standard library features.

````python
# Standard library module import
import math

# Module function usage with namespace qualification
radius = 5
area = math.pi * radius ** 2
print(f"Circle area (radius={radius}): {area:.2f}")
````

## **Section 21: Selective Function Importing**

Selective importing enables direct access to specific functions without namespace qualification, reducing code verbosity.

````python
# Import specific functions from math module
from math import pi, sqrt

# Direct function access without module prefix
area_direct = pi * radius ** 2
print(f"Circle area (direct access): {area_direct:.2f}")

# Square root calculation using imported function
print(f"Square root of 16: {sqrt(16)}")
````

## **Section 22: Module Aliasing**

Module aliasing provides convenient shorthand references for frequently used modules, particularly beneficial for modules with lengthy names.

````python
# Module import with alias assignment
import math as m

# Module usage through alias
area_alias = m.pi * radius ** 2
print(f"Circle area (aliased module): {area_alias:.2f}")
````

## **Section 23: Basic Exception Handling**

Exception handling prevents program termination from runtime errors, enabling graceful error recovery and user feedback.

````python
# Basic try-except structure for error handling
try:
    # Intentional error condition for demonstration
    numerator = 10
    denominator = 0
    result = numerator / denominator
    print("Result:", result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
````

## **Section 24: Multiple Exception Types**

Comprehensive exception handling addresses various error conditions through specific exception type matching and generic catching.

````python
# Function demonstrating multiple exception handling
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("ZeroDivisionError: Cannot divide by zero")
        return None
    except TypeError:
        print("TypeError: Invalid input types")
        return None
    except Exception as e:
        # Generic exception catching with error message
        print(f"Unexpected error: {e}")
        return None
    else:
        # Execution when no exception occurs
        print(f"Division successful: {result}")
        return result
    finally:
        # Always executed regardless of exception occurrence
        print("Division operation completed")

# Normal operation test case
safe_divide(10, 2)

# Zero division error test case
safe_divide(10, 0)

# Type error test case
safe_divide("10", 2)
````

## **Section 25: Class Definition and Constructor**

Object-oriented programming organises code through class definitions, encapsulating data and behaviour within structured entities.

````python
# Class definition with constructor method
class Dog:
    # Constructor method for object initialisation
    def __init__(self, name, age):
        # Instance attribute assignment
        self.name = name
        self.age = age

    # Instance method definition
    def bark(self):
        print("Woof! Woof!")
````

````python
# Object instantiation with constructor arguments
dog1 = Dog("Buddy", 3)
dog2 = Dog("Max", 5)

# Instance attribute access
print("Name:", dog1.name)
print("Age:", dog1.age)

# Instance method invocation
dog1.bark()
````

## **Section 26: Class Inheritance Structure**

Inheritance enables code reuse through parent-child class relationships, allowing specialisation whilst maintaining common functionality.

````python
# Base class definition
class Animal:
    # Parent class constructor
    def __init__(self, species):
        self.species = species

    # Method intended for overriding
    def make_sound(self):
        pass

# Child class inheriting from Animal
class Dog(Animal):
    # Child class constructor
    def __init__(self, name, age):
        # Parent constructor invocation
        super().__init__("Canine")
        self.name = name
        self.age = age

    # Method overriding from parent class
    def make_sound(self):
        print("Woof! Woof!")
````

````python
# Object creation using inherited class
dog = Dog("Buddy", 3)

# Access to inherited attribute from parent class
print("Species:", dog.species)

# Invocation of overridden method
dog.make_sound()
````
